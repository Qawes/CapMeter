# Kapacitás-idő diagram blokksorrenddel

Az STFT a `plot_stft=False` paraméterrel kikapcsolható. Az RMSD és a mozgásbesorolás a `rms_window_size`, `rms_threshold_right` és `rms_threshold_left` paraméterekkel jelenik meg. A `first_block_tasks` és `second_block_tasks` cellában a konzolra kiírt blokksorrendeket kell megadni.
