# Trim sleep data template

Edit the `TRIM_JOBS` list in the next cell with the recording date and the number of seconds to remove from the front and back of each FDC recording. Then run the cells below it.

The notebook calls the shared implementation in [`trim_sleep_data.py`](./trim_sleep_data.py), so the trimming behavior stays identical to the command-line script. Outputs are written to `data_analysis/sleep_tests/datasets_for_training` by default.

```python
from datetime import date
from pathlib import Path

from trim_sleep_data import (
    DEFAULT_DATA_DIRECTORY,
    DEFAULT_OUTPUT_DIRECTORY,
    parse_trim_jobs,
    trim_sleep_data,
)

# Replace these example values with the recordings you want to trim.
# front_trim and back_trim are measured in seconds.
TRIM_JOBS = [
    {"date": "2026-10-16", "front_trim": 900, "back_trim": 60},
    {"date": "2026-10-17", "front_trim": 3600, "back_trim": 40},
    {"date": "2026-10-19", "front_trim": 5400, "back_trim": 40},
    {"date": "2026-10-20", "front_trim": 400, "back_trim": 15000}
]

DATA_DIRECTORY = DEFAULT_DATA_DIRECTORY
OUTPUT_DIRECTORY = DEFAULT_OUTPUT_DIRECTORY

print("Data directory:", DATA_DIRECTORY)
print("Output directory:", OUTPUT_DIRECTORY)
```

```python
jobs = parse_trim_jobs(TRIM_JOBS)
print(f"Prepared {len(jobs)} trim job(s).")
```

```python
for recording_date, front_trim, back_trim in jobs:
    print(
        f"\n========== Trimming {recording_date} "
        f"(front={front_trim:g}s, back={back_trim:g}s) =========="
    )
    trim_sleep_data(
        recording_date,
        front_trim,
        back_trim,
        data_directory=Path(DATA_DIRECTORY),
        output_directory=Path(OUTPUT_DIRECTORY),
    )
```
